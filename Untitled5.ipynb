```python

import heapq
import matplotlib.pyplot as plt

# 1. Create the graph
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D', 'E'],
    'D': ['B', 'C', 'E'],
    'E': ['C', 'D']
}

# 2. Dijkstra's shortest path algorithm
def dijkstra(graph, start):
    distance = {}

    for node in graph:
        distance[node] = float('inf')

    distance[start] = 0
    pq = [(0, start)]

    while pq:
        current_distance, current_node = heapq.heappop(pq)

        if current_distance > distance[current_node]:
            continue

        for neighbour in graph[current_node]:
            new_distance = current_distance + 1

            if new_distance < distance[neighbour]:
                distance[neighbour] = new_distance
                heapq.heappush(pq, (new_distance, neighbour))

    return distance


# 3. Find shortest paths from B
start = 'B'
shortest_path = dijkstra(graph, start)

print("Shortest Path from", start)

for node in shortest_path:
    print(start, "->", node, "=", shortest_path[node])


# 4. Calculate network density
V = len(graph)
E = sum(len(graph[node]) for node in graph) // 2

density = (2 * E) / (V * (V - 1))

print("\nNumber of Vertices =", V)
print("Number of Edges =", E)
print("Density =", density)


# 5. Calculate degree of each node
degree = {}

for node in graph:
    degree[node] = len(graph[node])

print("\nDegree of each node:")

for node in degree:
    print(node, "=", degree[node])


# 6. Find the ego node
ego_node = max(degree, key=degree.get)

print("\nEgo Node =", ego_node)


# 7. Draw the graph manually (without NetworkX)
pos = {
    'A': (0, 2),
    'B': (1, 3),
    'C': (1, 1),
    'D': (2, 2),
    'E': (2, 0)
}

plt.figure(figsize=(6, 5))

# Draw edges
drawn_edges = set()

for node in graph:
    for neighbour in graph[node]:
        edge = tuple(sorted([node, neighbour]))

        if edge not in drawn_edges:
            drawn_edges.add(edge)

            x_values = [pos[node][0], pos[neighbour][0]]
            y_values = [pos[node][1], pos[neighbour][1]]

            plt.plot(x_values, y_values, color='gray', linewidth=1.5)

# Draw nodes
for node, (x, y) in pos.items():
    color = 'red' if node == ego_node else 'skyblue'

    plt.scatter(
        x, y,
        color=color,
        s=1200,
        edgecolors='black',
        linewidth=1.5,
        zorder=2
    )

    plt.text(
        x, y, node,
        fontsize=12,
        fontweight='bold',
        ha='center',
        va='center',
        zorder=3
    )

plt.title(f"Graph Network (Ego Node: {ego_node})")
plt.axis('off')
plt.show()
```
